**Base de datos:** Valor agregado del sector agropecuario como % del PBI, por país (2021)
**Fuente:** GitHub - argendatafundar/data

## 1. Cargar los datos

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('agro_global_b.csv')
df.head()
```

```python
df.info()
```

## 2. Medidas de tendencia central
Variable analizada: `va_agro_sobre_pbi` (% del PBI que representa el sector agropecuario)

```python
media = df['va_agro_sobre_pbi'].mean()
mediana = df['va_agro_sobre_pbi'].median()
moda = df['va_agro_sobre_pbi'].mode()[0]

print(f'Media: {media:.2f}%')
print(f'Mediana: {mediana:.2f}%')
print(f'Moda: {moda:.2f}%')
```

## 3. Medidas de dispersión

```python
desv_std = df['va_agro_sobre_pbi'].std()
varianza = df['va_agro_sobre_pbi'].var()
rango = df['va_agro_sobre_pbi'].max() - df['va_agro_sobre_pbi'].min()

print(f'Desviación estándar: {desv_std:.2f}')
print(f'Varianza: {varianza:.2f}')
print(f'Rango: {rango:.2f}')
print()
df['va_agro_sobre_pbi'].describe()
```

## 4. Gráficos

```python
plt.figure(figsize=(8,5))
df['va_agro_sobre_pbi'].hist(bins=20, color='steelblue', edgecolor='black')
plt.title('Distribución del valor agregado agropecuario (% del PBI)')
plt.xlabel('% del PBI')
plt.ylabel('Cantidad de países')
plt.show()
```

```python
plt.figure(figsize=(6,5))
df.boxplot(column='va_agro_sobre_pbi')
plt.title('Boxplot - % del PBI agropecuario por país')
plt.ylabel('% del PBI')
plt.show()
```

```python
top10 = df.sort_values('va_agro_sobre_pbi', ascending=False).head(10)

plt.figure(figsize=(9,5))
plt.barh(top10['geonombreFundar'], top10['va_agro_sobre_pbi'], color='seagreen')
plt.xlabel('% del PBI')
plt.title('Top 10 países con mayor peso agropecuario en el PBI (2021)')
plt.gca().invert_yaxis()
plt.show()
```
